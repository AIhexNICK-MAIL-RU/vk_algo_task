```python
def найти_все_позиции(строка, подстрока):
    """
    Находит все позиции вхождения подстроки в строку.
    Возвращает список индексов, где начинается каждое вхождение.
    """
    длины = {
        'строка': len(строка),
        'подстрока': len(подстрока)
    }
    print(f"Длина строки: {длины['строка']}")
    print(f"Длина подстроки: {длины['подстрока']}\n")

    позиции = []
    текущая_позиция = 0

    while текущая_позиция < длины['строка']:
        print(f"\nПроверяем позицию {текущая_позиция}:")

        # Проверяем совпадение начиная с текущей позиции
        совпадение = True
        for i in range(длины['подстрока']):
            if (текущая_позиция + i >= длины['строка'] or
                строка[текущая_позиция + i] != подстрока[i]):
                совпадение = False
                break

        if совпадение:
            print(f"Найдено совпадение на позиции {текущая_позиция}")
            позиции.append(текущая_позиция)
        else:
            print("Совпадение не найдено")

        текущая_позиция += 1

    return позиции

# Пример использования
строка = "алфавит алфавит алфавит"
подстрока = "алфа"
результат = найти_все_позиции(строка, подстрока)
print("\nВсе найденные позиции:", результат)
```

```python
def boyer_moore_horspool(text, pattern):
    """
    Реализация алгоритма поиска подстроки Boyer-Moore-Horspool.

    Args:
        text (str): Исходная строка
        pattern (str): Искомая подстрока

    Returns:
        int: Индекс начала найденной подстроки или -1 если подстрока не найдена
    """
    len_text = len(text)
    len_pattern = len(pattern)

    # Базовые проверки
    if len_pattern > len_text:
        return -1
    if len_pattern == 0:
        return 0 if len_text == 0 else -1

    # Создание таблицы смещений
    offset = {}
    for i in range(len_pattern - 1):
        offset[pattern[i]] = len_pattern - i - 1

    # Поиск
    skip_val = len_pattern - 1
    while skip_val <= len_text:
        match = True
        lastIndex = 0
        currentLetter = ''

        # Сравнение символов справа налево
        for i in range(len_pattern):
            lastIndex = i
            currentLetter = text[skip_val - 1 - i]
            if pattern[len_pattern - 1 - i] != text[skip_val - 1 - i]:
                match = False
                break

        if match:
            return skip_val - len_pattern

        # Вычисление смещения
        if lastIndex > 0:
            skip_val += offset.get(pattern[-1], len_pattern)
            continue
        skip_val += offset.get(currentLetter, len_pattern)

    return -1

# Тест 1: Базовый тест с простым случаем
def test_boyer_moore_horspool():
    text = "ABABABABC"
    pattern = "ABABC"
    result = boyer_moore_horspool(text, pattern)
    assert result == 4, f"Ожидалось 4, получено {result}"
    print("Тест 1 пройден успешно")

# Тест 2: Тест с несколькими вхождениями
def test_multiple_occurrences():
    text = "ABCABCABC"
    pattern = "ABC"
    result = boyer_moore_horspool(text, pattern)
    assert result == 0, f"Ожидалось 0, получено {result}"
    print("Тест 2 пройден успешно")

# Запуск тестов
test_boyer_moore_horspool()
test_multiple_occurrences()
```

```python
def hash(s):
    """
    Функция вычисления хеша строки.

    Args:
        s (str): Входная строка

    Returns:
        int: Хеш-значение строки
    """
    if not s:
        return 0

    result = ord(s[0])
    x = 31
    q = 2147483647

    for i in range(len(s) - 1):
        result = result * x + ord(s[i + 1])

    return result % q

def rabina_karpa(text, pattern):
    """
    Реализация алгоритма Рабина-Карпа для поиска подстроки.

    Args:
        text (str): Исходный текст
        pattern (str): Искомая подстрока

    Returns:
        int: Индекс начала найденной подстроки или None если не найдено
    """
    x = 31
    q = 2147483647
    pattern_hash = hash(pattern)
    m = len(pattern)
    current_hash = hash(text[0:m])
    i = 0

    while True:
        if pattern_hash == current_hash:
            if pattern == text[i:i + m]:
                return i
        if i + m >= len(text):
            break
        current_hash = ((current_hash - ord(text[i]) * x ** (m - 1)) * x + ord(text[i + m])) % q
        i = i + 1

    return None

# Тест 1: Базовый тест с простым случаем
def test_rabina_karpa():
    text = "ABABABABC"
    pattern = "ABABC"
    result = rabina_karpa(text, pattern)
    assert result == 4, f"Ожидалось 4, получено {result}"
    print("Тест 1 пройден успешно")

# Тест 2: Тест с несколькими вхождениями
def test_multiple_occurrences():
    text = "ABCABCABC"
    pattern = "ABC"
    result = rabina_karpa(text, pattern)
    assert result == 0, f"Ожидалось 0, получено {result}"
    print("Тест 2 пройден успешно")

# Запуск тестов
test_rabina_karpa()
test_multiple_occurrences()
```

Префикс - функция
● Префикс - строка P является префиксом для строки L если существует
некоторая строка X для которой выполняется равенство L = PX, где PX -
конкатенация двух строк. Иными словами это та часть с которой
начинается наша строка.
● P - подстрока для L с начальным индексом равным 0
● Собственный префикс - префикс называется собственным если X не
пустая строка. Иными словами, если длина префикса меньше длины
строки.
● Суффикс - строка S является суффиксом для строки L если
справедливо равенство L = XS. То есть эта та часть L на которую она
заканчивается.
● S - подстрока для L с конечным индексом n
● Собственный суффикс - если X не пустая строка, то есть если S
меньше L

```python
def prefix_function(text):
    """
    Функция вычисления префикс-функции для строки.
    Префикс-функция возвращает массив, где элемент на позиции i
    равен длине наибольшего собственного префикса строки text[0..i],
    который является также суффиксом этой строки.

    Args:
        text (str): Входная строка

    Returns:
        list: Массив значений префикс-функции
    """
    n = len(text)
    pi = [0] * n
    for i in range(1, n):
        j = pi[i - 1]
        while j > 0 and text[j] != text[i]:
            j = pi[j - 1]
        if text[i] == text[j]:
            j += 1
        pi[i] = j
    return pi

# Тест 1: Базовый тест с простой строкой
def test_prefix_function():
    text = "AABAAB"
    expected = [0, 1, 0, 1, 2, 3]
    result = prefix_function(text)
    assert result == expected, f"Ожидалось {expected}, получено {result}"
    print("Тест 1 пройден успешно")

# Тест 2: Тест со строкой с повторяющимися символами
def test_repeating_symbols():
    text = "AAAAA"
    expected = [0, 1, 2, 3, 4]
    result = prefix_function(text)
    assert result == expected, f"Ожидалось {expected}, получено {result}"
    print("Тест 2 пройден успешно")

# Запуск тестов
test_prefix_function()
test_repeating_symbols()
```

```python
def prefix_function(text):
    """
    Функция вычисления префикс-функции для строки.
    Префикс-функция возвращает массив, где элемент на позиции i
    равен длине наибольшего собственного префикса строки text[0..i],
    который является также суффиксом этой строки.

    Args:
        text (str): Входная строка

    Returns:
        list: Массив значений префикс-функции
    """
    n = len(text)
    pi = [0] * n
    for i in range(1, n):
        j = pi[i - 1]
        while j > 0 and text[j] != text[i]:
            j = pi[j - 1]
        if text[i] == text[j]:
            j += 1
        pi[i] = j
    return pi

def kmp_search(text, pattern, start_index=0):
    """
    Реализация алгоритма Кнута-Морриса-Пратта для поиска подстроки.

    Args:
        text (str): Исходный текст
        pattern (str): Искомая подстрока
        start_index (int): Начальный индекс для поиска

    Returns:
        int: Индекс начала найденной подстроки или None если не найдено
    """
    if not pattern:
        return 0 if not text else None
    if not text:
        return None
    if start_index >= len(text):
        return None
    if start_index < 0:
        return None

    j = 0
    pi = prefix_function(pattern)

    for i in range(start_index, len(text)):
        while j > 0 and text[i] != pattern[j]:
            j = pi[j - 1]
        if text[i] == pattern[j]:
            j += 1
        if j >= len(pattern):
            return i - j + 1
    return None

# Тест 1: Базовый тест с простым случаем
def test_kmp_search():
    text = "ABABABABC"
    pattern = "ABABC"
    result = kmp_search(text, pattern)
    assert result == 4, f"Ожидалось 4, получено {result}"
    print("Тест 1 пройден успешно")

# Тест 2: Тест с краевыми случаями
def test_edge_cases():
    # Пустые строки
    assert kmp_search("", "") == 0, "Пустые строки должны совпадать"
    assert kmp_search("abc", "") == None, "Паттерн не может быть пустым"
    assert kmp_search("", "abc") == None, "Текст не может быть пустым"

    # Начальный индекс
    assert kmp_search("ABCABC", "ABC", 2) == 3, "Поиск с начального индекса"
    assert kmp_search("ABCABC", "ABC", 6) == None, "Начальный индекс за пределами строки"

    print("Тест 2 пройден успешно")

# Запуск тестов
test_kmp_search()
test_edge_cases()
```

```python
def z_function(s):
    """
    Функция вычисления Z-функции для строки.
    Z-функция возвращает массив, где элемент на позиции i
    равен длине наибольшего префикса строки s[i:], который
    совпадает с префиксом всей строки s.

    Args:
        s (str): Входная строка

    Returns:
        list: Массив значений Z-функции
    """
    n = len(s)
    z = [0] * n
    l, r = 0, 0

    for i in range(1, n):
        # если мы находимся в границе z-блока
        if i <= r:
            # ищем уже рассчитанное значение z[i - l],
            # но не дальше правой границы
            z[i] = min(r - i + 1, z[i - l])

        # дальше каждое успешное увеличение z[i] сдвинет
        # z-блок на единицу
        while i + z[i] < n and s[z[i]] == s[i + z[i]]:
            z[i] += 1

        # проверим, правее ли мы текущего z-блока
        if i + z[i] - 1 > r:
            # устанавливаем новые границы
            r = i + z[i] - 1
            l = i

    return z

# Тест 1: Базовый тест с простой строкой
def test_z_function():
    text = "AABAAB"
    expected = [0, 1, 0, 3, 1, 0]
    result = z_function(text)
    assert result == expected, f"Ожидалось {expected}, получено {result}"
    print("Тест 1 пройден успешно")

# Тест 2: Тест с краевыми случаями
def test_edge_cases():
    # Пустая строка
    assert z_function("") == [], "Пустая строка должна вернуть пустой список"

    # Односимвольная строка
    assert z_function("A") == [1], "Односимвольная строка должна вернуть [1]"

    # Строка с повторяющимися символами
    assert z_function("AAAA") == [4, 3, 2, 1], "Правильная Z-функция для строки с повторами"

    print("Тест 2 пройден успешно")

# Запуск тестов
test_z_function()
test_edge_cases()
```
