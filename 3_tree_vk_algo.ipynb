```python
# Восстановление бинарного дерева из массива
# class TreeNode{
#    function TreeNode(val = 0, left = null, right = null) {
#        this.data = val
#        this.left = left
#        this.right = right
#    }
# }

# function buildTree(arr, i) { 
#    // определяем базовые кейсы
#    // создаем узел
#    // добавляем левого потомка 
#    // добавляем правого потомка

#    if i >= len(arr) {
#        return null
#    }

#    root = TreeNode(arr[i])
#    root.left = buildTree(arr, 2 * i + 1) 
#    root.right = buildTree(arr, 2 * i + 2)

#    return root
# }

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right

def build_tree(arr, i):
    # Базовый случай: если индекс выходит за границы массива, возвращаем None
    if i >= len(arr) or arr[i] is None:
        return None

    # Создаем новый узел с текущим значением
    root = TreeNode(arr[i])

    # Рекурсивно строим левое и правое поддерево
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)

    return root

# Функция для вывода дерева в порядке обхода Preorder
def print_tree_preorder(node):
    if not node:
        return []
    return [node.data] + print_tree_preorder(node.left) + print_tree_preorder(node.right)

# Тест 1: Простое бинарное дерево
arr1 = [1, 2, 3, 4, 5, 6, 7]
root1 = build_tree(arr1, 0)
print("Test Case 1 Output:", print_tree_preorder(root1))  # Ожидаемый результат: [1, 2, 4, 5, 3, 6, 7]

# Тест 2: Бинарное дерево с пропущенными узлами
arr2 = [1, 2, 3, None, 5, None, 7]
root2 = build_tree(arr2, 0)
print("Test Case 2 Output:", print_tree_preorder(root2))  # Ожидаемый результат: [1, 2, 5, 3, 7]
```

```python
# Симметричное дерево BFS 
# function isSymmetric(root) {
#    if root == null {
#        return true;
#    }
#    queue = [root];

#    while (len(queue) > 0) { 
#        queueLen = len(queue);
#        for (i = 0; i < queueLen; i++) { 
#            //queue[queueLen - i - 1] - симметричный 
#     //относительно узла queue[i]
#            if (queue[i] == null && queue[queueLen - i - 1] == null) {
#                continue;
#            }
#            if (queue[i] == null || queue[queueLen - i - 1] == null) {
#                return false;
#            }
#            if (queue[i].data != queue[queueLen - i - 1].data) {
#                return false;
#            }

#            queue.append(queue[i].left);
#            queue.append(queue[i].right);
#        }
#        queue = queue[queueLen:]; 
#    }

#    return true;
# }

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right

def is_symmetric(root: TreeNode) -> bool:
    if not root:
        return True
        
    queue = deque([root])
    
    while queue:
        level_size = len(queue)
        current_level = []
        
        # Собираем узлы текущего уровня
        for _ in range(level_size):
            node = queue.popleft()
            current_level.append(node.data if node else None)
            
            if node:
                queue.append(node.left)
                queue.append(node.right)
        
        # Проверяем симметричность текущего уровня
        for i in range(level_size // 2):
            if current_level[i] != current_level[level_size - 1 - i]:
                return False
                
    return True

# Тестовые случаи
def test_symmetric_tree():
    # Симметричное дерево
    #        1
    #     /     \
    #    2       2
    #   / \     / \
    #  3   4   4   3
    root = TreeNode(1)
    root.left = TreeNode(2, TreeNode(3), TreeNode(4))
    root.right = TreeNode(2, TreeNode(4), TreeNode(3))
    assert is_symmetric(root) == True

def test_asymmetric_tree():
    # Несимметричное дерево
    #        1
    #     /     \
    #    2       2
    #     \       \
    #      3       3
    root = TreeNode(1)
    root.left = TreeNode(2, None, TreeNode(3))
    root.right = TreeNode(2, None, TreeNode(3))
    assert is_symmetric(root) == False

def test_empty_tree():
    assert is_symmetric(None) == True

def test_single_node():
    assert is_symmetric(TreeNode(5)) == True

# Запуск тестов
test_symmetric_tree()
test_asymmetric_tree()
test_empty_tree()
test_single_node()
print("All tests passed!")
```

```python
# Симметричное дерево DFS
# function isSymmetricDFS(root) {
#    if root == null {
#        return true
#    }
#    data = []
#    data = deptSearch(root, data)
#    j = len(data) - 1
#    for i = 0; i < len(data)/2; i++ {
#        if data[i] != data[j] {
#            return false
#        }
#        j--
#    }

#    return true
# }

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right

def is_symmetric_dfs(root: TreeNode) -> bool:
    if not root:
        return True

    def is_mirror(left, right):
        if not left and not right:
            return True
        if not left or not right:
            return False
        return (left.data == right.data) and is_mirror(left.left, right.right) and is_mirror(left.right, right.left)

    return is_mirror(root.left, root.right)

# Тест 1: Симметричное дерево (должно пройти)
#     1
#   /   \
#  2     2
def test_symmetric_simple():
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(2)
    assert is_symmetric_dfs(root) == True

# Тест 2: Несимметричное дерево (должно упасть)
#     1
#   /   \
#  2     3
def test_asymmetric():
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    assert is_symmetric_dfs(root) == False

# Тест 3: Пустое дерево (должно пройти)
def test_empty_tree():
    assert is_symmetric_dfs(None) == True

# Тест 4: Одиночный узел (должно пройти)
def test_single_node():
    assert is_symmetric_dfs(TreeNode(5)) == True

# Запуск тестов
test_symmetric_simple()
test_asymmetric()
test_empty_tree()
test_single_node()
print("Все тесты пройдены!")
```

```python
# Поиск минимальной глубины бинарного дерева DFS
# function minDepth(root) {

#    if root == null {

#        return 0

#    }

#    if root.left == nil && root.right == nil {

#      return 1

#    }

#    if root.left != null && root.right != null {

#        return 1 + min(minDepth(root.left), minDepth(root.right))

#    }

#    if root.left != null {

#        return 1 + minDepth(root.left)

#    }

#    if root.right != null {

#        return 1 + minDepth(root.right)

#    }

# }
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

def minDepth(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 1
    if root.left is not None and root.right is not None:
        return 1 + min(minDepth(root.left), minDepth(root.right))
    if root.left is not None:
        return 1 + minDepth(root.left)
    if root.right is not None:
        return 1 + minDepth(root.right)

# Тест 1: Бинарное дерево с двумя листьями на одинаковом уровне
def test_min_depth_equal_leaves():
    # Создание дерева:
    #     1
    #    / \
    #   2   3
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    assert minDepth(root) == 2

# Тест 2: Бинарное дерево с листьями на разных уровнях
def test_min_depth_unequal_leaves():
    # Создание дерева:
    #     1
    #    /
    #   2
    #  /
    # 3
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.left.left = TreeNode(3)
    assert minDepth(root) == 3

test_min_depth_equal_leaves()
test_min_depth_unequal_leaves()
```

```python
# Поиск максимального произведения
# function maxMinMultiplication(data) {
#    if length(data) < 3 {
#        return -1
#    }

#    min_index = 1 
#    max_index = 2 
#    for i = 1; i < length(data); i = 2 * i + 1 {
#        min_index = i
#    }

#    for i = 1; i < length(data); i = 2 * i + 2 {
#        max_index = i
#    }

#    result = data[min_index] * data[max_index] 
#    return result
# }

def maxMinMultiplication(data):
    if len(data) < 3:
        return -1

    # Сортировка массива
    sorted_data = sorted(data)

    # Максимальное произведение может быть либо произведением двух максимальных чисел,
    # либо произведением двух минимальных чисел (если они отрицательные)
    max_product = max(sorted_data[-1] * sorted_data[-2], sorted_data[0] * sorted_data[1])

    return max_product

# Тест 1: Массив с положительными числами
def test_max_min_multiplication_positive():
    data = [1, 2, 3, 4, 5]
    assert maxMinMultiplication(data) == 20

# Тест 2: Массив с отрицательными и положительными числами
def test_max_min_multiplication_mixed():
    data = [-10, -5, 0, 1, 2]
    assert maxMinMultiplication(data) == 50

test_max_min_multiplication_positive()
test_max_min_multiplication_mixed()
```

```python
# Сравнение двух деревьев
# function isSubtree(A, B) {
#    // Пустое дерево B считается поддеревом
#    // любого дерева 
#    if (B == null) {
#        return true;
#    }
#    // Если A пусто, а B нет — не может быть поддеревом 
#    if (A == null) {
#        return false;
#    }
#    if (isSameTree(A, B)) {
#        return true;
#    }
#    // Рекурсивно проверяем левое и правое поддеревья A 
#    return isSubtree(A.left, B) || isSubtree(A.right, B);
# }
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

def isSameTree(A, B):
    if A is None and B is None:
        return True
    if A is None or B is None:
        return False
    if A.val != B.val:
        return False
    return isSameTree(A.left, B.left) and isSameTree(A.right, B.right)

def isSubtree(A, B):
    # Пустое дерево B считается поддеревом любого дерева
    if B is None:
        return True
    # Если A пусто, а B нет — не может быть поддеревом
    if A is None:
        return False
    if isSameTree(A, B):
        return True
    # Рекурсивно проверяем левое и правое поддеревья A
    return isSubtree(A.left, B) or isSubtree(A.right, B)

# Тест 1: Дерево B является поддеревом дерева A
def test_is_subtree_true():
    # Создание дерева A:
    #     1
    #    / \
    #   2   3
    #  /
    # 4
    A = TreeNode(1)
    A.left = TreeNode(2)
    A.right = TreeNode(3)
    A.left.left = TreeNode(4)
    
    # Создание дерева B:
    #     2
    #    /
    #   4
    B = TreeNode(2)
    B.left = TreeNode(4)
    
    assert isSubtree(A, B) == True

# Тест 2: Дерево B не является поддеревом дерева A
def test_is_subtree_false():
    # Создание дерева A:
    #     1
    #    / \
    #   2   3
    A = TreeNode(1)
    A.left = TreeNode(2)
    A.right = TreeNode(3)
    
    # Создание дерева B:
    #     4
    B = TreeNode(4)
    
    assert isSubtree(A, B) == False

test_is_subtree_true()
test_is_subtree_false()
```

```python
# Зеркальные узлы
# function dfs(left, right) {
#    if left == nil || right == nil {
#       return 0
#    }

#    count = 0
#    // Нашли пару
#    if left.val == right.val {
#       count = 1
#    }

#    // Проверяем детей в зеркальном порядке 
#    count += dfs(left.left, right.right) 
#    count += dfs(left.right, right.left)

#    return count
# }
# func CountMirrorTwins(root) {
#    if root == nil {
#       return 0
#    }
#    return dfs(root.left, root.right)
# }
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

def dfs(left, right):
    if left is None or right is None:
        return 0

    count = 0
    # Нашли пару
    if left.val == right.val:
        count = 1

    # Проверяем детей в зеркальном порядке
    count += dfs(left.left, right.right)
    count += dfs(left.right, right.left)

    return count

def count_mirror_twins(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 0
    return dfs(root.left, root.right)

# Тест 1: Дерево с зеркальными узлами
def test_count_mirror_twins_true():
    # Создание дерева:
    #     1
    #    / \
    #   2   2
    #  / \ / \
    # 3  4 4  3
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(2)
    root.left.left = TreeNode(3)
    root.left.right = TreeNode(4)
    root.right.left = TreeNode(4)
    root.right.right = TreeNode(3)
    
    assert count_mirror_twins(root) == 3


# Тест 2: Дерево без зеркальных узлов
def test_count_mirror_twins_false():
    # Создание дерева:
    #     1
    #    / \
    #   2   3
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    
    assert count_mirror_twins(root) == 0

```
