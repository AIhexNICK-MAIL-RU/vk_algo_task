```python
# Найти корень числа (ближайшее целое)
# function binarySearchSqrt(target) {
#    l = 0; r = target
#    while l <= r {
#        middle = (l+r) / 2
#        if middle2 > target {
#            r = middle-1
#            continue
#        }
#        if middle2 < target {
#            l = middle+1
#            continue
#        }
#        return middle
#    }
#    return r
# }
def binary_search_sqrt(target):
    l = 0
    r = target
    while l <= r:
        middle = (l + r) / 2
        if middle * middle > target:
            r = middle - 1e-9  # Уменьшаем правую границу (учёт погрешности для вещественных чисел)
        elif middle * middle < target:
            l = middle + 1e-9  # Увеличиваем левую границу
        else:
            return middle
    return r  # Возвращаем приближённое значение

# Тест 1: Проверка квадратного корня из 16 (ожидается 4)
result1 = binary_search_sqrt(16)
print(f"√16 ≈ {result1:.6f}")  # Ожидаем ~4.000000

# Тест 2: Проверка квадратного корня из 2 (ожидается ~1.414213)
result2 = binary_search_sqrt(2)
print(f"√2 ≈ {result2:.6f}")  # Ожидаем ~1.414213
```

```python
# Очень лёгкая задача
# function copyTime(n, x, y) {
#    l = 0
#    r = (n - 1) * max(x, y)

#    while l + 1 < r {
#        mid = (r + l) / 2
#        if mid/x + mid/y < n-1 {
#            l = mid
#        } else {
#            r = mid
#        }
#    }

#    return r + min(x, y)
# }
def copy_time(n, x, y):
    if n == 1:
        return min(x, y)
    
    left = 0
    right = (n - 1) * max(x, y)
    
    while left + 1 < right:
        mid = (left + right) // 2  # Используем целочисленное деление для дискретного случая
        if (mid // x) + (mid // y) < n - 1:
            left = mid
        else:
            right = mid
    
    return right + min(x, y)

# Тест 1: 1 копия (ожидается min(x, y))
print(copy_time(1, 2, 5))  # Ожидаем 2

# Тест 2: 5 копий (x=2, y=5 → ожидается 9)
print(copy_time(5, 2, 5))  # Ожидаем 9
```

```python
# Накормить животных
# function feedAnimals(animals, food) {
#     if len(animals) == 0 or len(food) == 0 {
#         return 0
#     }
#     sort(animals) // animals.sort()
#     sort(food)    // food.sort()

#     count = 0
#     for f in food {
#         if f >= animals[count] {
#             count += 1
#         }

#         if count == len(animals) {
#             break
#         }
#     }

#     return count
# }
def feed_animals(animals, food):
    if len(animals) == 0 or len(food) == 0:
        return 0
    
    animals_sorted = sorted(animals)
    food_sorted = sorted(food)
    
    count = 0
    for f in food_sorted:
        if f >= animals_sorted[count]:
            count += 1
        
        if count == len(animals_sorted):
            break
    
    return count

# Тест 1: Пустые списки (ожидается 0)
print(feed_animals([], []))  # Ожидаем 0

# Тест 2: Обычный случай (3 животных можно накормить)
print(feed_animals([3, 2, 1], [1, 2, 3, 4]))  # Ожидаем 3
```

```python
# Найти разницу между двух строк
# function extraLetter(a, b) {
# 	hashMapA := map[string]int{}
# 	// O(n)
# 	for i := 0; i < len(a); i++ {
# 		hashMapA[a[i]]++
# 	}
# 	// O(n)
# 	for i := 0; i < len(b); i++ {
# 		if contains(hashMapA, b[i]) {
# 			hashMapA[b[i]]--
# 			// как только счетчик становится
# 			// равен 0 - удаляем элемент
# 			// из хеш-таблицы
# 			if hashMapA[b[i]] == 0 {
# 				delete(hashMapA, b[i])
# 				continue
# 			}
# 			continue
# 		}
#                           // если мы удалили ключ со значением b[i]
#                           // или же этой буквы никогда не было в таблицы
# 		return b[i]
# 	}

# 	return ""
# }
def extra_letter(a: str, b: str) -> str:
    from collections import defaultdict

    # Создаем словарь для подсчета символов в строке a
    count = defaultdict(int)
    for char in a:
        count[char] += 1

    # Проверяем каждый символ в строке b
    for char in b:
        if char in count:
            count[char] -= 1
            if count[char] == 0:
                del count[char]
        else:
            return char

    # Если все символы совпали, возвращаем пустую строку
    return ""

# Тест 1: Одна лишняя буква в середине
print(extra_letter("abcd", "abecd"))  # Ожидаем 'e'

# Тест 2: Лишняя буква в конце
print(extra_letter("abc", "abcz"))  # Ожидаем 'z'
```

```python
# Сумма двух элементов
# function twoSum(data, target) {
# 	cache = map[int]int
# 	for i = 0; i < len(data); i++ {
# 		cache[data[i]] = i
# 	}
# 	for i = 0; i < len(data); i++ {
# 		diff = target - data[i]
# 		if contains(cache, diff) {
# 		    return [i, cache[diff]]
# 		}
# 	}

# 	return []
# }
def two_sum(data, target):
    cache = {}
    # Сначала заполняем кеш (значение -> индекс)
    for i in range(len(data)):
        cache[data[i]] = i
    
    # Проверяем для каждого элемента, есть ли (target - элемент) в кеше
    for i in range(len(data)):
        diff = target - data[i]
        if diff in cache and cache[diff] != i:  # Убедимся, что это не тот же элемент
            return [i, cache[diff]]
    
    return []  # Если пара не найдена

# Тест 1: Обычный случай (найдена пара)
print(two_sum([2, 7, 11, 15], 9))  # Ожидаем [0, 1] (так как 2 + 7 = 9)

# Тест 2: Нет подходящей пары
print(two_sum([3, 5, 8], 10))  # Ожидаем [] (нет пары, дающей 10)
```

```python
# Сортировка шелла
# function shell_sort(arr) {
#     n = len(arr)
#     gap = len(arr) / 2
#     while gap > 0 {
#         for current_position = gap; current_position < n; current_position++ {
#             m_gap = current_position
#             while m_gap >= gap and arr[m_gap] < arr[m_gap - gap] {
#                 arr[m_gap], arr[m_gap - gap] = arr[m_gap - gap], arr[m_gap]
#                 m_gap -= gap
#             }
#         }
#         gap = gap / 2
#     }
#     return arr
# }
def shell_sort(arr):
    n = len(arr)
    gap = n // 2  # Используем целочисленное деление
    
    while gap > 0:
        for current_position in range(gap, n):
            m_gap = current_position
            # Сдвигаем элемент на правильную позицию в текущем подмассиве
            while m_gap >= gap and arr[m_gap] < arr[m_gap - gap]:
                # Меняем элементы местами
                arr[m_gap], arr[m_gap - gap] = arr[m_gap - gap], arr[m_gap]
                m_gap -= gap
        gap = gap // 2  # Уменьшаем шаг
    
    return arr

# Тест 1: Сортировка случайного массива
test1 = [12, 34, 54, 2, 3]
print("До сортировки:", test1)
print("После сортировки:", shell_sort(test1.copy()))
# Ожидаемый результат: [2, 3, 12, 34, 54]

# Тест 2: Сортировка уже отсортированного массива
test2 = [1, 2, 3, 4, 5]
print("\nДо сортировки:", test2)
print("После сортировки:", shell_sort(test2.copy()))
# Ожидаемый результат: [1, 2, 3, 4, 5]
```

```python
# Дан массив строк strs. Сгруппируйте анаграммы вместе. Вы можете вернуть ответ в любом порядке.

# Анаграмма - это слово или фраза, полученная путем перестановки букв другого слова или фразы, обычно с использованием всех исходных букв ровно один раз.

# Входные данные: ["eat","tea","tan","ate","nat","bat"]
# Выходные: [["bat"],["nat","tan"],["ate","eat","tea"]]

# Входные данные: ["won","now","aaa","ooo","ooo"]
# Выходные: [["aaa"],["ooo", "ooo"],["won","now"]]
from collections import defaultdict

def group_anagrams(strs):
    anagrams = defaultdict(list)
    
    for s in strs:
        # Используем отсортированную строку как ключ
        sorted_s = ''.join(sorted(s))
        anagrams[sorted_s].append(s)
    
    return list(anagrams.values())

# Тест 1
input1 = ["eat","tea","tan","ate","nat","bat"]
print(group_anagrams(input1))
# Ожидаемый результат: [["bat"],["nat","tan"],["ate","eat","tea"]]

# Тест 2
input2 = ["won","now","aaa","ooo","ooo"]
print(group_anagrams(input2))
# Ожидаемый результат: [["aaa"],["ooo","ooo"],["won","now"]]
```
