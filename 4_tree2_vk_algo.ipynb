```python
# Проверка корректности кучи
# function isMaxHeap(root) {
#    if (!root) { 
#        return true;
#    }

#    // Очередь для обхода дерева в ширину (BFS) 
#    queue = [root];
#    shouldBeLeaf = false;

#    while (queue.length > 0) { 
#        current = queue.pop();

#        // Проверяем левый потомок 
#        if (current.left) {
#            if (shouldBeLeaf || current.left.val > current.val) {
#                return false;
#            }
#            queue.push(current.left);
#        } else {
#            shouldBeLeaf = true; // Все следующие узлы должны быть листьями 
#        }

#        // Проверяем правый потомок 
#        if (current.right) {
#            if (shouldBeLeaf || current.right.val > current.val) {
#                return false;
#            }
#            queue.push(current.right);
#        } else {
#            shouldBeLeaf = true; // Все следующие узлы должны быть листьями 
#        }
#    }

#    return true;
# }

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_max_heap(root):
    if not root:
        return True

    # Очередь для обхода дерева в ширину (BFS)
    queue = deque([root])
    should_be_leaf = False

    while queue:
        current = queue.popleft()

        # Проверяем левый потомок
        if current.left:
            if should_be_leaf or current.left.val > current.val:
                return False
            queue.append(current.left)
        else:
            should_be_leaf = True  # Все следующие узлы должны быть листьями

        # Проверяем правый потомок
        if current.right:
            if should_be_leaf or current.right.val > current.val:
                return False
            queue.append(current.right)
        else:
            should_be_leaf = True  # Все следующие узлы должны быть листьями

    return True

# Тест 1: Корректная максимальная куча
#     5
#    / \
#   3   4
#  / \
# 1   2
def test_valid_max_heap():
    # Создание дерева
    root = TreeNode(5)
    root.left = TreeNode(3)
    root.right = TreeNode(4)
    root.left.left = TreeNode(1)
    root.left.right = TreeNode(2)

    assert is_max_heap(root) == True

# Тест 2: Некорректная максимальная куча
#     3
#    / \
#   5   4
def test_invalid_max_heap():
    # Создание дерева
    root = TreeNode(3)
    root.left = TreeNode(5)
    root.right = TreeNode(4)

    assert is_max_heap(root) == False

test_valid_max_heap()
test_invalid_max_heap()
```

```python
# Полное бинарное дерево
#     if (!root) { 
#         return true;
#     }
#     queue = [root]; 
#     seenNull = false;
#     while (queue.length > 0) {
#         node = queue.pop();
#         if (!node) {
#             seenNull = true;
#         } else {
#             if (seenNull) {
#                 return false;
#             }
#             queue.push(node.left);
#             queue.push(node.right);
#         }
#     }
#     return true;
# }

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_complete_binary_tree(root):
    if not root:
        return True

    queue = deque([root])
    seen_null = False

    while queue:
        node = queue.popleft()
        if not node:
            seen_null = True
        else:
            if seen_null:
                return False
            queue.append(node.left)
            queue.append(node.right)

    return True

# Тест 1: Полное бинарное дерево
#     1
#    / \
#   2   3
#  / \ / \
# 4  5 6  7
def test_complete_tree():
    # Создание дерева
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    root.left.left = TreeNode(4)
    root.left.right = TreeNode(5)
    root.right.left = TreeNode(6)
    root.right.right = TreeNode(7)

    assert is_complete_binary_tree(root) == True

# Тест 2: Неполное бинарное дерево
#     1
#    / \
#   2   None
#  /
# 3
def test_incomplete_tree():
    # Создание дерева
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.left.left = TreeNode(3)

    assert is_complete_binary_tree(root) == False


test_complete_tree()
test_incomplete_tree()
```

```python
# Объединение K отсортированных массивов

# function mergeKSortedArrays(sortedArrays) {
#    // Итоговый массив 
#    mergedArray = [];

#    // Минимальная куча 
#    minHeap = new MinHeap();

#    // Инициализируем кучу с первым элементом каждого массива 
#    for (i = 0; i < len(sortedArrays); i++) {
#        currentArray = sortedArrays[i];
#        // Проверяем, что массив не пустой 
#        if (len(currentArray) > 0) {
#            // (значение, индекс массива, индекс элемента) 
#            minHeap.heappush([currentArray[0], i, 0]);
#        }
#    }

#    // Пока куча не пуста
#    while (!minHeap.isEmpty()) {
#        // Извлекаем наименьший элемент из кучи
#        value, arrayIdx, elementIdx = minHeap.heappop(); 
#        // Добавляем его в итоговый массив 
#        mergedArray.append(value);

#        // Проверяем, есть ли следующий элемент в том же массиве 
#        if (elementIdx + 1 < len(sortedArrays[arrayIdx])) {
#            // Добавляем следующий элемент этого массива в кучу 
#            nextElement = sortedArrays[arrayIdx][elementIdx + 1]; 
#            minHeap.push([nextElement, arrayIdx, elementIdx + 1]);
#        }
#    }

#    return mergedArray;
# }

import heapq

def merge_k_sorted_arrays(sorted_arrays):
    # Итоговый массив
    merged_array = []

    # Минимальная куча
    min_heap = []

    # Инициализируем кучу с первым элементом каждого массива
    for i, current_array in enumerate(sorted_arrays):
        # Проверяем, что массив не пустой
        if current_array:
            # (значение, индекс массива, индекс элемента)
            heapq.heappush(min_heap, (current_array[0], i, 0))

    # Пока куча не пуста
    while min_heap:
        # Извлекаем наименьший элемент из кучи
        value, array_idx, element_idx = heapq.heappop(min_heap)
        # Добавляем его в итоговый массив
        merged_array.append(value)

        # Проверяем, есть ли следующий элемент в том же массиве
        if element_idx + 1 < len(sorted_arrays[array_idx]):
            # Добавляем следующий элемент этого массива в кучу
            next_element = sorted_arrays[array_idx][element_idx + 1]
            heapq.heappush(min_heap, (next_element, array_idx, element_idx + 1))

    return merged_array

# Тест 1: Объединение трех отсортированных массивов
def test_merge_arrays():
    sorted_arrays = [
        [1, 3, 5],
        [2, 4, 6],
        [0, 7, 8]
    ]
    expected_result = [0, 1, 2, 3, 4, 5, 6, 7, 8]
    assert merge_k_sorted_arrays(sorted_arrays) == expected_result

# Тест 2: Объединение двух отсортированных массивов
def test_merge_two_arrays():
    sorted_arrays = [
        [1, 3, 5],
        [2, 4, 6]
    ]
    expected_result = [1, 2, 3, 4, 5, 6]
    assert merge_k_sorted_arrays(sorted_arrays) == expected_result

test_merge_arrays()
test_merge_two_arrays()
```

```python
# К-ый наименьший/наибольший элемент в BST

# function inorderMin(node, k, counter) {
#    if (node == null) {
#        return null;
#    }

#    // Обход левого поддерева
#    leftResult = inorderMin(node.left, k, counter); 
#    if (leftResult != null) {
#         // Если нашли результат, возвращаем его 
#   return leftResult; 
#    }

#    // Увеличиваем счетчик 
#    counter++;

#    // Проверяем, является ли текущий узел k-м наименьшим 
#    if (counter == k) {
#        return node.val;
#    }

#    // Обход правого поддерева
#    return inorderMin(node.right, k, counter);
# }

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def inorder_min(node, k, counter=[0]):
    if node is None:
        return None

    # Обход левого поддерева
    left_result = inorder_min(node.left, k, counter)
    if left_result is not None:
        # Если нашли результат, возвращаем его
        return left_result

    # Увеличиваем счетчик
    counter[0] += 1

    # Проверяем, является ли текущий узел k-м наименьшим
    if counter[0] == k:
        return node.val

    # Обход правого поддерева
    return inorder_min(node.right, k, counter)

def inorder_max(node, k, counter=[0]):
    if node is None:
        return None

    # Обход правого поддерева
    right_result = inorder_max(node.right, k, counter)
    if right_result is not None:
        # Если нашли результат, возвращаем его
        return right_result

    # Увеличиваем счетчик
    counter[0] += 1

    # Проверяем, является ли текущий узел k-м наибольшим
    if counter[0] == k:
        return node.val

    # Обход левого поддерева
    return inorder_max(node.left, k, counter)

# Тест 1: Поиск k-го наименьшего элемента
#     4
#    / \
#   2   6
#  / \   \
# 1   3   5
def test_kth_min():
    # Создание дерева
    root = TreeNode(4)
    root.left = TreeNode(2)
    root.right = TreeNode(6)
    root.left.left = TreeNode(1)
    root.left.right = TreeNode(3)
    root.right.right = TreeNode(5)

    assert inorder_min(root, 3) == 3

# Тест 2: Поиск k-го наибольшего элемента
def test_kth_max():
    # Создание дерева
    root = TreeNode(4)
    root.left = TreeNode(2)
    root.right = TreeNode(6)
    root.left.left = TreeNode(1)
    root.left.right = TreeNode(3)
    root.right.right = TreeNode(5)

    assert inorder_max(root, 3) == 4

test_kth_min()
test_kth_max()
```

```python
# Balance factor

# function calculateHeightsAndBalance (node) {
#    if (!node) {
#        return 0; // Высота пустого дерева равна 0
#    }

#    // Рекурсивно вычисляем высоты левого и правого поддеревьев 
#  leftHeight = calculateHeightsAndBalance (node.left); 
#  rightHeight = calculateHeightsAndBalance (node.right);

#    // Проставляем баланс-фактор текущего узла 
#    node.balanceFactor = leftHeight - rightHeight;

#    // Возвращаем высоту текущего поддерева 
#    return 1 + max(leftHeight, rightHeight);
# }

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        self.balanceFactor = None

def calculate_heights_and_balance(node):
    if node is None:
        return 0  # Высота пустого дерева равна 0

    # Рекурсивно вычисляем высоты левого и правого поддеревьев
    left_height = calculate_heights_and_balance(node.left)
    right_height = calculate_heights_and_balance(node.right)

    # Проставляем баланс-фактор текущего узла
    node.balanceFactor = left_height - right_height

    # Возвращаем высоту текущего поддерева
    return 1 + max(left_height, right_height)

# Тест 1: Высота и баланс-фактор для сбалансированного дерева
#     1
#    / \
#   2   3
def test_balanced_tree():
    # Создание дерева
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)

    calculate_heights_and_balance(root)

    # Проверка высот и баланс-факторов
    assert root.balanceFactor == 0
    assert root.left.balanceFactor == 0
    assert root.right.balanceFactor == 0

# Тест 2: Высота и баланс-фактор для несбалансированного дерева
#     1
#    /
#   2
#  /
# 3
def test_unbalanced_tree():
    # Создание дерева
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.left.left = TreeNode(3)

    calculate_heights_and_balance(root)

    # Проверка высот и баланс-факторов
    assert root.balanceFactor == 2
    assert root.left.balanceFactor == 1
    assert root.left.left.balanceFactor == 0

test_balanced_tree()
test_unbalanced_tree()
```

```python
# Преобразование в зеркальное дерево

# function mirrorTreeIterative(node) {
#    if (node === null) { 
#        return null;
#    }

#    // Очередь для обхода в ширину (BFS) 
#    queue = [node];

#    while (len(queue) > 0) { 
#        current = queue.pop();

#        temp = current.left;
#        current.left = current.right; 
#        current.right = temp;

#        // Добавляем в очередь дочерние узлы 
#        if (current.left) {
#            queue.push(current.left);
#        }
#        if (current.right) {
#            queue.push(current.right);
#        }
#    }

#    return node;
# }

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def mirror_tree_iterative(root):
    if root is None:
        return None

    # Очередь для обхода в ширину (BFS)
    queue = deque([root])

    while queue:
        current = queue.popleft()

        # Меняем местами левого и правого детей
        current.left, current.right = current.right, current.left

        # Добавляем в очередь дочерние узлы
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)

    return root

# Тест 1: Преобразование дерева в зеркальное
#     1
#    / \
#   2   3
#  /
# 4
def test_mirror_tree():
    # Создание исходного дерева
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    root.left.left = TreeNode(4)

    # Преобразование в зеркальное
    mirror_root = mirror_tree_iterative(root)

    # Проверка структуры зеркального дерева
    assert mirror_root.val == 1
    assert mirror_root.left.val == 3
    assert mirror_root.right.val == 2
    assert mirror_root.right.left is None
    assert mirror_root.right.right.val == 4

# Тест 2: Преобразование пустого дерева
def test_mirror_empty_tree():
    assert mirror_tree_iterative(None) is None

test_mirror_tree()
test_mirror_empty_tree()
```
