```python
# Структура узла в боре
# CLASS TrieNode:
#     METHOD Constructor():
#         children = {}  // Словарь для хранения дочерних узлов
#         is_end_of_word = FALSE  // Флаг, указывающий, является ли узел концом слова

class TrieNode:
    def __init__(self):
        """
        Конструктор узла префиксного дерева (Trie)
        """
        self.children = {}  # Словарь для хранения дочерних узлов (ключ - символ, значение - узел)
        self.is_end_of_word = False  # Флаг, указывающий конец слова

# Пример использования
if __name__ == "__main__":
    # Создаем корневой узел
    root = TrieNode()
    
    # Добавляем слово "apple"
    current = root
    for char in "apple":
        if char not in current.children:
            current.children[char] = TrieNode()
        current = current.children[char]
    current.is_end_of_word = True
    
    # Проверяем наличие слова
    def search(word):
        current = root
        for char in word:
            if char not in current.children:
                return False
            current = current.children[char]
        return current.is_end_of_word
    
    print(search("apple"))  # True
    print(search("app"))    # False
```

```python
# Поиск (псевдокод)
# FUNCTION search(word):
#     current_node = root
#     FOR EACH char IN word:
#         IF char NOT IN current_node.children:
#             RETURN FALSE
#         current_node = current_node.children[char]
#     RETURN current_node.is_end_of_word
class TrieNode:
    def __init__(self):
        self.children = {}  # Хранит дочерние узлы (ключ - символ, значение - узел)
        self.is_end_of_word = False  # Флаг конца слова

class Trie:
    def __init__(self):
        self.root = TrieNode()  # Создаем корневой узел
    
    def insert(self, word):
        """
        Вставляет слово в префиксное дерево
        :param word: строка для вставки
        :return: None
        """
        current_node = self.root
        for char in word:
            # Если символа нет в дочерних узлах - создаем новый узел
            if char not in current_node.children:
                current_node.children[char] = TrieNode()
            # Переходим к следующему узлу
            current_node = current_node.children[char]
        # Помечаем последний узел как конец слова
        current_node.is_end_of_word = True

# Пример использования
trie = Trie()
words = ["apple", "app", "application", "banana"]

print("Вставка слов:")
for word in words:
    print(f"Добавляем: '{word}'")
    trie.insert(word)

# Проверка содержимого
def print_trie(node, prefix=""):
    if node.is_end_of_word:
        print(f"Найдено слово: {prefix}")
    for char, child in node.children.items():
        print_trie(child, prefix + char)

print("\nСодержимое Trie:")
print_trie(trie.root)
```

```python
# Поиск по префиксу (псевдокод)
# FUNCTION searchPrefix(word):
#     current_node = root
#     FOR EACH char IN word:
#         IF char NOT IN current_node.children:
#             RETURN FALSE
#         current_node = current_node.children[char]
#     RETURN TRUE
class TrieNode:
    def __init__(self):
        self.children = {}  # Дочерние узлы
        self.is_end_of_word = False  # Флаг конца слова

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        """Вставляет слово в Trie"""
        current = self.root
        for char in word:
            if char not in current.children:
                current.children[char] = TrieNode()
            current = current.children[char]
        current.is_end_of_word = True
    
    def searchPrefix(self, prefix):
        """
        Проверяет, существует ли префикс в Trie
        :param prefix: искомый префикс
        :return: True если префикс существует, иначе False
        """
        current = self.root
        for char in prefix:
            if char not in current.children:
                return False
            current = current.children[char]
        return True
    
    def searchWord(self, word):
        """Проверяет, существует ли слово целиком"""
        current = self.root
        for char in word:
            if char not in current.children:
                return False
            current = current.children[char]
        return current.is_end_of_word

# Пример использования
trie = Trie()
words = ["apple", "app", "application", "banana"]

# Вставляем слова
for word in words:
    trie.insert(word)

# Тестируем searchPrefix
test_cases = [
    ("app", True),
    ("ban", True),
    ("orange", False),
    ("applic", True),
    ("apple", True)
]

print("Тестирование searchPrefix:")
for prefix, expected in test_cases:
    result = trie.searchPrefix(prefix)
    print(f"'{prefix}': {'✅' if result == expected else '❌'} (ожидалось {expected}, получили {result})")

# Дополнительная проверка searchWord
print("\nТестирование searchWord:")
print("'app':", trie.searchWord("app"))    # True
print("'apps':", trie.searchWord("apps"))  # False
```

```python
# Удаление (псевдокод)
# FUNCTION delete(word):
#     current_node = root
#     stack = EMPTY_STACK()
#     FOR EACH char IN word:
#         IF char NOT IN current_node.children:
#             RETURN  // Word not found, nothing to delete
#         stack.PUSH((current_node, char))  // Save the path
#         current_node = current_node.children[char]
#     IF NOT current_node.is_end_of_word:
#         RETURN  // Word not found, nothing to delete
#     SET current_node.is_end_of_word TO FALSE  // Mark as not end of word

#     IF current_node.children IS EMPTY:
#         WHILE stack IS NOT EMPTY:
#             (parent_node, char) = stack.POP()
#             DELETE parent_node.children[char]
#             IF parent_node.is_end_of_word OR parent_node.children IS NOT EMPTY:
#                 BREAK  // Stop if parent is end of another word or has other children

class TrieNode:
    def __init__(self):
        self.children = {}  # Дочерние узлы
        self.is_end_of_word = False  # Флаг конца слова

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        current = self.root
        for char in word:
            if char not in current.children:
                current.children[char] = TrieNode()
            current = current.children[char]
        current.is_end_of_word = True
    
    def search(self, word):
        current = self.root
        for char in word:
            if char not in current.children:
                return False
            current = current.children[char]
        return current.is_end_of_word
    
    def delete(self, word):
        """
        Удаляет слово из Trie
        :param word: слово для удаления
        :return: None
        """
        current = self.root
        stack = []  # Для хранения пути
        
        # Шаг 1: Поиск слова и сохранение пути
        for char in word:
            if char not in current.children:
                return  # Слово не найдено
            stack.append((current, char))
            current = current.children[char]
        
        if not current.is_end_of_word:
            return  # Слово не найдено (только префикс)
        
        # Шаг 2: Удаление отметки конца слова
        current.is_end_of_word = False
        
        # Шаг 3: Удаление лишних узлов (если нет дочерних и не конец другого слова)
        if not current.children:
            while stack:
                parent_node, char = stack.pop()
                del parent_node.children[char]
                
                # Прекращаем если:
                # 1. Родитель является концом другого слова
                # 2. У родителя есть другие дочерние узлы
                if parent_node.is_end_of_word or parent_node.children:
                    break

# Пример использования
trie = Trie()
words = ["apple", "app", "application", "banana"]

# Вставляем слова
for word in words:
    trie.insert(word)

# Проверяем перед удалением
print("До удаления:")
print("'app' exists:", trie.search("app"))
print("'apple' exists:", trie.search("apple"))

# Удаляем слово
trie.delete("apple")

# Проверяем после удаления
print("\nПосле удаления 'apple':")
print("'app' exists:", trie.search("app"))      # Должно остаться
print("'apple' exists:", trie.search("apple"))  # Должно быть удалено

# Удаляем префикс
trie.delete("app")

print("\nПосле удаления 'app':")
print("'app' exists:", trie.search("app"))      # Удалено
print("'application' exists:", trie.search("application"))  # Должно остаться
```

```python
# Структура узла 
# CLASS RadixTreeNode:
#     CONSTRUCTOR(prefix = ""):
#         SET self.prefix TO prefix  // Префикс, хранящийся в узле
#         SET self.children TO {}    // Дочерние узлы (словарь)
#         SET self.is_end TO FALSE  // Флаг, указывающий, что узел завершает ключ

class RadixTree:
    def __init__(self):
        self.root = RadixTreeNode()
    
    def insert(self, key):
        """Вставка ключа в дерево"""
        current = self.root
        i = 0
        
        while i < len(key):
            # Поиск подходящего дочернего узла
            found = False
            for child_char, child_node in current.children.items():
                if child_char == key[i]:
                    # Нашли совпадение - идем глубже
                    current = child_node
                    i += 1
                    found = True
                    break
            
            if not found:
                # Создаем новый узел для оставшейся части ключа
                new_node = RadixTreeNode(prefix=key[i:])
                new_node.is_end = True
                current.children[key[i]] = new_node
                return
    
        # Помечаем последний узел как конец ключа
        current.is_end = True
    
    def search(self, key):
        """Поиск ключа в дереве"""
        current = self.root
        i = 0
        
        while i < len(key):
            found = False
            for child_char, child_node in current.children.items():
                if child_char == key[i]:
                    # Проверяем весь префикс дочернего узла
                    if key.startswith(child_node.prefix, i):
                        current = child_node
                        i += len(child_node.prefix)
                        found = True
                        break
                    else:
                        return False
            
            if not found:
                return False
        
        return current.is_end

# Пример использования
if __name__ == "__main__":
    tree = RadixTree()
    
    # Вставляем ключи
    keys = ["romane", "romanus", "romulus", "rubens", "ruber", "rubicon", "rubicundus"]
    for key in keys:
        tree.insert(key)
        print(f"Добавлено: {key}")
    
    # Проверяем наличие ключей
    test_cases = ["romane", "romanus", "rubicon", "ruby", "r"]
    for test in test_cases:
        print(f"'{test}': {'Найден' if tree.search(test) else 'Не найден'}")
```
